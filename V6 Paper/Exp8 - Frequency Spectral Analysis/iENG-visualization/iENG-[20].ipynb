```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import scipy
```

```python
import os
import h5py

path = 'D:/Dataset/The dataset/2020-02-18 Human Trial [20]/Data/'
f_lst = os.listdir(path)

for f in f_lst[:10]:
    file_path = os.path.join(path, f)
    print(f"\n--- {f} ---")

    with h5py.File(file_path, 'r') as mat:
        #print("Keys:", list(mat.keys()))

        def load_hdf5(obj):
            if isinstance(obj, h5py.Dataset):
                return obj[()]
            elif isinstance(obj, h5py.Group):
                out = {}
                for k, v in obj.items():
                    out[k] = load_hdf5(v)
                return out

        data = {k: load_hdf5(v) for k, v in mat.items()}

        plt.figure(figsize=(12, 4))
        plt.plot(data['Data_ADC0'].T[1000:])
        plt.show()

    print("Loaded variables:", list(data.keys()))
```
