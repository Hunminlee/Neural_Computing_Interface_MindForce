Protocols - two subjects
==============================
0. Without anything
- Model initialization
- Day 1. Test
- Day 2. Test
- ...
- Day 10. Test

1. Model retraining (Existing method)
- Initialization
- Day 1. Model adaptation -> Test
- Day 2. Model adaptation -> Test
- Day 3. Model adaptation -> Test
- …
- Day 10. Model adaptation -> Test

2. Human adaptation
- Initialization
- Day 1. Human trial (practice 5 mins) -> Test
- Day 2. Human trial (practice 5 mins) -> Test
- Day 3. Human trial (practice 5 mins) -> Test
- …
- Day 10. Human trial (practice 5 mins) -> Test

3. Human + AI co-adaptation
- Initialization
- Day 1. Model adaptation + human trial (first adaptation, then human practice) -> Test
- Day 2. Model adaptation + human trial (first adaptation, then human practice) -> Test
- Day 3. Model adaptation + human trial (first adaptation, then human practice) -> Test
- …
- Day 10. Model adaptation + human trial (first adaptation, then human practice) -> Test

```python
colors = {
    "Initialized Only": "#000000",   # Black
    "AI Only": "#0072B2",   # Blue
    "Human Only": "#009E73",   # Green
    "AI-Human Co-Adaptation": "#D55E00",   # Orange-Red
}

conditions = ["Initialized Only", "AI Only", "Human Only", "AI-Human Co-Adaptation"]

X = np.arange(1, 11)  # Days 1–10

plt.figure(figsize=(10, 6))

for cond in conditions:
    # Shape: (num_subjects, 10 days)
    data = np.array([subj[cond] for subj in subjects])

    mean = data.mean(axis=0)
    sd = data.std(axis=0)
    sem = sd / np.sqrt(data.shape[0])

    # ---- Option A: Use SEM (recommended for Nature) ----
    plt.errorbar(X, mean, yerr=sem, fmt='o-', color=colors[cond],
                 capsize=4, linewidth=2, markersize=6, label=cond)

    # ---- Option B (optional): use SD instead of SEM ----
    # plt.errorbar(X, mean, yerr=sd, ...)

plt.xlabel("Day", fontsize=14)
plt.ylabel("Decoding Accuracy (%)", fontsize=14)
plt.xticks(X)
plt.title("Performance Across Days (Mean ± SEM Across Subjects)", fontsize=15)
plt.legend(frameon=False, fontsize=12)

plt.grid(alpha=0.15)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import sem, t

# -------------------------------------------------
# Nature-style parameters
# -------------------------------------------------
plt.rcParams.update({
    "font.size": 14,
    "axes.linewidth": 1.2,
    "xtick.major.width": 1,
    "ytick.major.width": 1,
    "xtick.direction": "out",
    "ytick.direction": "out",
    "figure.dpi": 300,
})

colors = {
    "Initialized Only": "#000000",   # Black
    "AI Only": "#0072B2",   # Blue
    "Human Only": "#009E73",   # Green
    "AI-Human Co-Adaptation": "#D55E00",   # Orange-Red
}

conditions = ["Initialized Only", "AI Only", "Human Only", "AI-Human Co-Adaptation"]
X = np.arange(1, 11)

# -------------------------------------------------
# Function: Compute 95% CI
# -------------------------------------------------
def ci95(data):
    n = data.shape[0]
    se = sem(data, axis=0)
    h = se * t.ppf((1 + 0.95) / 2., n - 1)
    return h

# -------------------------------------------------
# Plot
# -------------------------------------------------
plt.figure(figsize=(9, 5))

for cond in conditions:
    # shape: (subjects, 10 days)
    data = np.array([s[cond] for s in subjects])

    mean = data.mean(axis=0)
    h = ci95(data)

    # Ribbon (95% CI)
    plt.fill_between(X, mean - h, mean + h, color=colors[cond], alpha=0.18, linewidth=0)

    # Main line (thick, clean)
    plt.plot(X, mean, color=colors[cond], linewidth=2.4, label=cond)

    # Markers (minimalistic)
    plt.scatter(X, mean, color=colors[cond], s=35, zorder=10)

# -------------------------------------------------
# Cosmetics
# -------------------------------------------------
plt.xlabel("Day", fontsize=15)
plt.ylabel("Accuracy (%)", fontsize=15)
plt.xticks(X)
plt.ylim(0, 100)

plt.legend(frameon=False, fontsize=12, ncol=4, bbox_to_anchor=(0.5, 1.12), loc="center")

# Cleaner Nature-style axes
plt.grid(alpha=0.08)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))
days = np.arange(1,11)

for cond in conditions:
    data = np.array([subj[cond] for subj in subjects])
    mean = data.mean(axis=0)
    sem = data.std(axis=0) / np.sqrt(len(subjects))

    # Plot per-subject faint lines
    for subj_data in data:
        plt.plot(days, subj_data, color=colors[cond], alpha=0.15, linewidth=1)

    # Group mean
    plt.plot(days, mean, color=colors[cond], linewidth=3)

    # Ribbon
    plt.fill_between(days, mean - sem, mean + sem,
                     color=colors[cond], alpha=0.18, label=cond)

plt.xlabel("Day")
plt.ylabel("Decoding Accuracy (%)")
plt.xticks(days)
plt.title("Subject-Level and Group-Level Performance", fontsize=16)
plt.legend(frameon=False)
plt.grid(alpha=0.1)
plt.tight_layout()
plt.show()
```
