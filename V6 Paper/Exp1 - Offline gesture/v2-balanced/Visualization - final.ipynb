```python
from itertools import zip_longest

def mean_std_ignore_none(*args):
    vals = [x for x in args if x is not None]
    if vals:
        return np.mean(vals), np.std(vals)
    else:
        return np.nan, np.nan

# Compute mean & std across Targets
mean_targets, std_targets = zip(*[
    mean_std_ignore_none(*vals)
    for vals in zip_longest(*Targets, fillvalue=None)
])

# Compute mean & std across Subs
mean_subs, std_subs = zip(*[
    mean_std_ignore_none(*vals)
    for vals in zip_longest(*Subs, fillvalue=None)
])

mean_targets, std_targets = np.array(mean_targets), np.array(std_targets)
mean_subs, std_subs = np.array(mean_subs), np.array(std_subs)

#plt.errorbar(range(len(mean_targets)), mean_targets, yerr=std_targets, label="Targets", fmt='-o', capsize=3)
#plt.errorbar(range(len(mean_subs)), mean_subs, yerr=std_subs, label="Subs", fmt='-o', capsize=3)

x_labels= [f'Visit {i}' for i in range(1,21)]
#plt.plot(x_labels, mean_targets, label="Targets", marker='o')
#plt.plot(x_labels, mean_subs, label="Subs", marker='o')

plt.plot(x_labels, np.mean(Targets, axis=0), label="Targets", marker='o')
plt.plot(x_labels, np.mean(Subs, axis=0), label="Subs", marker='o')

plt.xticks(rotation=45)  # Rotate labels by 45 degrees
plt.legend()
plt.tight_layout()       # Adjust layout so labels don’t get cut off
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import zip_longest

x_labels = [f'Visit {i}' for i in range(1, 22)]

# Example lists (yours should already exist)
Targets = [target_B, target_C, target_H2, target_M, target_X]
Subs = [subject_B, subject_C, subject_H2, subject_M, subject_X]

def collect_per_visit(data):
    return [
        [x for x in vals if x is not None]
        for vals in zip_longest(*data, fillvalue=None)
    ]

# Collect per-visit data separately
per_visit_targets = collect_per_visit(Targets)
per_visit_subs = collect_per_visit(Subs)

plt.figure(figsize=(14, 6), dpi=350)

positions_targets = np.arange(1, len(x_labels) + 1) - 0.2  # shift left
positions_subs = np.arange(1, len(x_labels) + 1) + 0.2     # shift right

# Boxplots
bp_targets = plt.boxplot(per_visit_targets, positions=positions_targets, widths=0.35, patch_artist=True,
            boxprops=dict(facecolor="lightblue"), medianprops=dict(color="blue"))
bp_subs = plt.boxplot(per_visit_subs, positions=positions_subs, widths=0.35, patch_artist=True,
            boxprops=dict(facecolor="lightcoral"), medianprops=dict(color="red"))

#plt.xticks(np.arange(1, len(x_labels) + 1), x_labels, rotation=45, fontsize=13)
plt.xticks(np.arange(1, len(x_labels) + 1), range(1, 22), rotation=0, fontsize=13)
plt.yticks(fontsize=13)
plt.xlabel("Visit number", fontsize=16)
plt.ylabel("Accuracy", fontsize=16)

# Fix legend: use the box patches as handles
handles = [bp_targets["boxes"][0], bp_subs["boxes"][0]]

labels = [
    "Intra-session: Target performance using random split",
    "Inter-session: Training on sessions 1…n and testing on the unseen n+1 session"
]
plt.legend(handles, labels, fontsize=15)

plt.tight_layout()
plt.show()
```
