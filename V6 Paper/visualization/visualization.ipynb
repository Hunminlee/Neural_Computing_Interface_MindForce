```python
import numpy as np
import matplotlib.pyplot as plt

# Number of samples
n_samples = 200

# Create four random 1D signals
signals = [np.random.randn(n_samples) for _ in range(4)]

# Colors for each channel
colors = ['r', 'g', 'b', 'm']

# Create subplots
fig, axes = plt.subplots(4, 1, figsize=(12, 7), sharex=True)

for i, ax in enumerate(axes):
    ax.plot(signals[i], color=colors[i])
    ax.set_ylabel(f'Ch {i+1}', fontsize=25)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_xticks([])  # remove x-axis ticks
    ax.set_yticks([])  # remove y-axis ticks

#plt.suptitle("Random 1D Signals for 4 Channels", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.7])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Reset style to default (white background)
plt.style.use('default')

# Number of samples
n_samples = 200

# Create four random 1D signals
signals = [np.random.randn(n_samples) for _ in range(4)]

# Colors for each channel
colors = ['r', 'g', 'b', 'm']

# Create subplots
fig, axes = plt.subplots(4, 1, figsize=(12, 7), sharex=True, facecolor='white')

for ax, sig, color in zip(axes, signals, colors):
    ax.plot(sig, color=color)
    ax.set_xticks([])  # remove x-axis ticks
    ax.set_yticks([])  # remove y-axis ticks
    ax.set_ylabel("")  # erase y-label
    ax.set_facecolor("white")  # ensure plot area is white
    ax.grid(False)             # remove grid

    # Remove plot edges (spines)
    for spine in ax.spines.values():
        spine.set_visible(False)

plt.tight_layout(rect=[0, 0, 1, 0.7])
plt.show()
```
